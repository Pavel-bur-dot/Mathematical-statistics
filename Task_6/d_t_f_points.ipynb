```python
import numpy as np
from scipy import stats

theta_true = 3
n = 100
alpha = 0.05
z = stats.norm.ppf(1 - alpha/2)

# Генерация выборки из данного распределения
# Метод обратной функции: X = (1 - U)^{-1/(θ-1)}
U = np.random.uniform(0, 1, n)
X = (1 - U) ** (-1/(theta_true - 1))

# Статистики
S = np.sum(np.log(X))
theta_hat = 1 + n / S

print(f"Истинное θ: {theta_true}")
print(f"Оценка ОМП: θ̂ = {theta_hat:.4f}")

print("\n-------------------------------пункт d------------------------------")
# Интервал для медианы
chi2_low = stats.chi2.ppf(alpha/2, 2*n)
chi2_high = stats.chi2.ppf(1 - alpha/2, 2*n)
m_low = 2 ** (2*S / chi2_high)
m_high = 2 ** (2*S / chi2_low)
m_true = 2 ** (1/(theta_true - 1))

print(f"\nМедиана:")
print(f"  Значение медианы: {m_true:.4f}")
print(f"  Доверительный иинтервал: [{m_low:.4f}, {m_high:.4f}]")

# Асимптотический интервал для θ
se_asymp = (theta_hat - 1) / np.sqrt(n)
theta_low = theta_hat - z * se_asymp
theta_high = theta_hat + z * se_asymp

print(f"\nАсимптотический доверительный интервал для θ:")
print(f"  [{theta_low:.4f}, {theta_high:.4f}]")
```

```python
# Параметрический бутстрап
n_bootstrap = 1000
theta_boot_par = []

for _ in range(n_bootstrap):
    # Генерируем выборку из Парето с оцененным θ
    U_boot = np.random.uniform(0, 1, n)
    X_boot = (1 - U_boot) ** (-1/(theta_hat - 1))
    S_boot = np.sum(np.log(X_boot))
    theta_boot_par.append(1 + n / S_boot)

left_par = np.percentile(theta_boot_par, 100 * alpha/2)
right_par = np.percentile(theta_boot_par, 100 * (1 - alpha/2))

# Непараметрический бутстрап
theta_boot_nonpar = []

for _ in range(n_bootstrap):
    X_boot = np.random.choice(X, n, replace=True)
    S_boot = np.sum(np.log(X_boot))
    theta_boot_nonpar.append(1 + n / S_boot)

left_nonpar = np.percentile(theta_boot_nonpar, 100 * alpha/2)
right_nonpar = np.percentile(theta_boot_nonpar, 100 * (1 - alpha/2))
print("\n--------------------------------пункт t------------------------------")
print(f"\nПараметрический бутстрап:")
print(f"  [{left_par:.4f}, {right_par:.4f}]")
print(f"\nНепараметрический бутстрап:")
print(f"  [{left_nonpar:.4f}, {right_nonpar:.4f}]")
```

```python
print("\n--------------------------пункт f--------------------------")

intervals = {
    "Асимптотический": (theta_low, theta_high),
    "Параметрический бутстрап": (left_par, right_par),
    "Непараметрический бутстрап": (left_nonpar, right_nonpar),
}

print(f"\n{'Метод':<30} {'левая гр':>10} {'правая гр':>10} {'Длина':>10}")
print("-"*65)

for name, (low, high) in intervals.items():
    length = high - low
    print(f"{name:<30} {low:>10.4f} {high:>10.4f} {length:>10.4f}")

print(f"\nИстинное значение θ = {theta_true}")
print(f"Оценка ОМП θ̂ = {theta_hat:.4f}")
```
