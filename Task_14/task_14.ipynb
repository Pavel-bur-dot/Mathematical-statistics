```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
```

```python
alpha = 0.05

quant = norm.ppf(1 - alpha)
print(f"Квантиль порядка {1-alpha} для N(0, 1): {quant:.4f}")

x = np.array([-1.11, -6.1, 2.42])
y = np.array([-2.29, -2.91])

x_av = np.mean(x)
y_av = np.mean(y)

delta = (x_av - y_av) / np.sqrt(7/6)

print(f"x_average = {x_av}; y_average = {y_av}; delta = {delta}")
```

```python
def W(tetha):
    return 1 - norm.cdf(1.645 - tetha / np.sqrt(7/6))

theta_values = np.linspace(0, 10, 500)

# Вычисление W
W_values = []
for theta in theta_values:
    val = W(theta)
    W_values.append(val)

# Построение графика
plt.figure(figsize=(10, 6))
plt.plot(theta_values, W_values, 'b-', linewidth=2)


plt.xlabel('θ', fontsize=12)
plt.ylabel('W', fontsize=12)
plt.grid(True, alpha=0.5)

plt.tight_layout()
```
