В данном файле представлен вспомогательный код с вычислением необходимых переменных

```python
import math
from scipy.stats import chi2

l = 0.61
n = 200
def P(k): 
    p = (l ** k) / (math.factorial(k)) * math.exp(-l)
    print(f"P{k} = {p}")
    return p


p = [0] * 5
print ("-------Веротности от ~lambda~")
for k in range(5):
    p[k] = P(k)
print("\n ------------n*P_i---------------")
for k in range(5):
    E = p[k]*n
    print(E)
print("\n-----------------P_i после объединения---------------")
p_new = [0] * 3
p_new[0] = p[0]
p_new[1] = p[1]
p_new[2] = 1 -(p_new[0] + p_new[1])

for k in range (3):
    print(p_new[k])

m = [109, 65, 26]
delta = 0
for k in range(3):
    delta += ((m[k] - n * p_new[k]) ** 2) / (n * p_new[k])

print(f"\n delta = {delta}")

df = 1

result = chi2.sf(delta, df)

print(f"({result} > 0.05) ")
```
