```python
from scipy.stats import f
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Параметры распределения
n = 139
s1_length = 5.722
s1_width = 4.612

m = 1000
s2_length = 6.161
s2_width = 5.055

x_length = s1_length ** 2 / s2_length ** 2
x_width = s1_width ** 2 / s2_width ** 2

alpha = 0.05

# Вычисление интеграла
F_left = f.ppf(alpha/2, n-1, m-1)
F_right = f.ppf(1 - alpha/2, n-1, m-1)

print(f"delta_длина = {x_length}")
print(f"delta_ширина = {x_width}")

print(f"G: delta <= {F_left} or delta > {F_right}")
print("Нет попаданий в G => нет оснований отвергать H_0")
```

Далее блок с построением графика мощности.

```python
def W(tetha):
    return f.cdf(0.77 / tetha, n-1, m-1) + 1 - (f.cdf(1.27 / tetha, n-1, m-1))

theta_values = np.linspace(0.1, 3, 500)

W_values = []
for theta in theta_values:
    val = W(theta)
    W_values.append(val)

# Построение графика
plt.figure(figsize=(8, 5))
plt.plot(theta_values, W_values, 'b-', linewidth=2, label=f'F({n-1}, {m-1})')

plt.xlabel('θ', fontsize=12)
plt.ylabel('W', fontsize=12)
plt.grid(True, alpha=0.5)
plt.legend()

plt.legend()
plt.tight_layout()
```
