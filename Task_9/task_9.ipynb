В данном файле представлен вспомогательный код с вычислением необходимых переменных

```python
from scipy.stats import chi2

data = [[33, 43, 80, 144],
        [39, 35, 75, 154]]
n = 600

p = [0]*4
p[0] = (33+39)/n
p[1] = (43+35)/n
p[2] = (80+75)/n
p[3] = (144+154)/n

for k in range(4):
    print(f"p_{k} = {p[k]}")
print("\n------------n*P_i:-----------\n")
for k in range(4):
    print(n*p[k])

delta_1 = 0
for j in range (4):
    delta_1 += (data[0][j]- 300*p[j])**2/(300*p[j])
print(f"\ndelta_1 = {delta_1}")

delta_2 = 0
for j in range (4):
    delta_2 += (data[1][j]- 300*p[j])**2/(300*p[j])
print(f"delta_2 = {delta_2}")

delta = delta_1 + delta_2
print(f"    delta = {delta}")

df = (4 - 1) * (2 - 1)
p_value = chi2.sf(delta, df)

print(f"p-value = {p_value:.6f} > 0,05")
```
