В данном файле представлен вспомогательный код с вычислением необходимых переменных

```python
from scipy.stats import chi2

part_1 = [25, 50, 25]
part_2 = [52, 41, 7]
n = 200

p = [0]*3
for k in range(3):
    p[k] = (part_1[k]+part_2[k])/n
    print (f"p{k} = {p[k]}")

q = [0.5]*2
print(f"q1 = {q[0]} and q2 = {q[1]}")

table = [[25, 50, 25],
        [52, 41, 7]]

delta = 0
for i in range(2):
    for j in range(3):
        delta += (table[i][j]-n*p[j]*q[i])**2/(n*p[j]*q[i])

print(f"\n      delta = {delta}")
df = (2-1)*(3-1)
print(f"\ndf = {df}")

result = chi2.sf(delta, df)
print(f"({result:.6f} < 0.05) ")
```
